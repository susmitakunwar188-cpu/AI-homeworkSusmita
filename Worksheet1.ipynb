```python
#4.1
#PROBLEM-1: ARRAY CREATION

import numpy as np

empty_array = np.empty((2,2))
print(empty_array)

ones_array = np.ones((4,2))
print(ones_array)

filled_array = np.full((3,3), 7)
print(filled_array)

zeros_like_array = np.zeros_like(filled_array)
print(zeros_like_array)

ones_like_array = np.ones_like(filled_array)
print(ones_like_array)

new_list = [1,2,3,4]
array_from_list = np.array(new_list)
print(array_from_list)
```

```python
#PROBLEM-2: ARRAY MANIPULATION
import numpy as np

arr_10_49 = np.arange(10,50)
print(arr_10_49)

matrix_0_8 = np.arange(9).reshape((3,3))
print(matrix_0_8)

identity_matrix = np.eye(3)
print(identity_matrix)

rand_array_30 = np.random.random(30)
print(rand_array_30)
print(rand_array_30.mean())

rand_10x10 = np.random.random((10,10))
print(rand_10x10)
print(rand_10x10.min(), rand_10x10.max())

zero_10 = np.zeros(10)
zero_10[4] = 1
print(zero_10)

arr_to_reverse = np.array([1,2,0,0,4,0])
reversed_arr = arr_to_reverse[::-1]
print(reversed_arr)

border_array = np.ones((5,5))
border_array[1:-1,1:-1] = 0
print(border_array)

checkerboard = np.zeros((8,8))
checkerboard[1::2, ::2] = 1
checkerboard[::2, 1::2] = 1
print(checkerboard)
```

```python
#PROBLEM-3: ARRAY OPERATION
import numpy as np

x = np.array([[1,2],[3,5]])
y = np.array([[5,6],[7,8]])
v = np.array([9,10])
w = np.array([11,12])

print(x + y)
print(x - y)
print(x * 2)
print(x**2)
print(np.dot(v,w))
print(np.dot(x,v))
print(np.dot(x,y))
print(np.concatenate((x,y), axis=0))
print(np.vstack((v,w)))
print("Cannot concatenate x and v directly because shapes are different.")
```

```python
#PROBLEM-4: MATRIX OPERATION
import numpy as np

A = np.array([[3,4],[7,8]])
B = np.array([[5,3],[2,1]])

A_inv = np.linalg.inv(A)
print(np.dot(A, A_inv))

AB = np.dot(A,B)
BA = np.dot(B,A)
print(AB)
print(BA)

AB_T = (AB).T
B_TA_T = np.dot(B.T, A.T)
print(AB_T)
print(B_TA_T)

coeff = np.array([[2,-3,1],[1,-1,2],[3,1,-1]])
const = np.array([-1,-3,9])
solution = np.dot(np.linalg.inv(coeff), const)
print(solution)
```

```python
#4.2
#1. Element-wise Addition:
import numpy as np
import time

size = 1000000
list1 = list(range(size))
list2 = list(range(size))

# Python list addition
start = time.time()
add_list = [list1[i] + list2[i] for i in range(size)]
end = time.time()
print("Python list addition time:", end-start)

# Numpy array addition
arr1 = np.array(list1)
arr2 = np.array(list2)
start = time.time()
add_np = arr1 + arr2
end = time.time()
print("Numpy array addition time:", end-start)
```

```python
#3. Element-wise Multiplication:
import numpy as np
import time

size = 1000000
list1 = list(range(size))
list2 = list(range(size))


start = time.time()
mult_list = [list1[i] * list2[i] for i in range(size)]
end = time.time()
print("Python list multiplication time:", end-start)

arr1 = np.array(list1)
arr2 = np.array(list2)
start = time.time()
mult_np = arr1 * arr2
end = time.time()
print("Numpy array multiplication time:", end-start)
```

```python
#3. Dot Product
import numpy as np
import time

size = 1000000
list1 = list(range(size))
list2 = list(range(size))

# Python list dot product
start = time.time()
dot_list = sum([list1[i] * list2[i] for i in range(size)])
end = time.time()
print("Python list dot product time:", end-start)

# Numpy array dot product
arr1 = np.array(list1)
arr2 = np.array(list2)
start = time.time()
dot_np = np.dot(arr1, arr2)
end = time.time()
print("Numpy array dot product time:", end-start)
```

```python
#4. Matrix Multiplication
import numpy as np
import time

size_mat = 1000

# Python lists
mat1 = [[i for i in range(size_mat)] for j in range(size_mat)]
mat2 = [[i for i in range(size_mat)] for j in range(size_mat)]
start = time.time()
res_list = [[sum(mat1[i][k]*mat2[k][j] for k in range(size_mat)) for j in range(size_mat)] for i in range(size_mat)]
end = time.time()
print("Python list matrix multiplication time:", end-start)

# Numpy arrays
mat_np1 = np.random.random((size_mat, size_mat))
mat_np2 = np.random.random((size_mat, size_mat))
start = time.time()
res_np = np.dot(mat_np1, mat_np2)
end = time.time()
print("Numpy array matrix multiplication time:", end-start)
```
